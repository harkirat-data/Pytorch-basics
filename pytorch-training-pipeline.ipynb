```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.shape
```

```python
df.drop(['id','Unnamed: 32'],inplace=True,axis=1)
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:],df.iloc[:,0],test_size = 0.2)
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_train
```

```python
y_train
```

```python
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

# Numpy arrays to Pytorch Tensors

```python
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test)
y_train_tensor = torch.from_numpy(y_train)
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor.shape
```

```python
y_train_tensor.shape
```

# Defining the Model

```python
import torch.nn as nn

class MyNN(nn.Module):

  def __init__(self, num):
    super().__init__()
    self.linear = nn.Linear(num,1)
    self.sigmoid = nn.Sigmoid()

  def forward(self,features):
    out = self.linear(features)
    out = self.sigmoid(out)

    return out

```

# Important Params

```python
learning_rate = 0.1
epochs = 25
```

```python
loss_function = nn.BCELoss()
```

# Training Pipeline

```python
model = MyNN(X_train_tensor.shape[1]) # create model

# Convert tensors to float32 to match model weights dtype
X_train_float = X_train_tensor.float()
y_train_float = y_train_tensor.float()

optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)

#define loop

for epoch in range(epochs):

  #forward pass
  y_pred = model(X_train_float)

  #calculate loss
  loss = loss_function(y_pred, y_train_float.view(-1,1))

  #clear grad

  optimizer.zero_grad()

  #backward pass
  loss.backward()

  #params update
  optimizer.step()

  #print loss in each epoch
  print(f'Epoch: {epoch+1}, Loss: {loss.item()}')
```

# Evaluation

```python
with torch.no_grad():
  # Convert inputs to float to match the model weights
  y_pred = model.forward(X_test_tensor.float())
  y_pred = (y_pred > 0.5).float()

  # Convert target to float and match the shape
  y_test_float = y_test_tensor.float().view(-1, 1)

  accuracy = (y_pred == y_test_float).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
