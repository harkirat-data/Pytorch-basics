```python
#tokenize
def tokenize(text):
  text = text.lower()
  text = text.replace('?','')
  text = text.replace('.','')
  text = text.replace(',','')
  text = text.replace('!','')
  return text.split()
```

```python
tokenize('What is the capital of France?')
```

```python
#vocab
vocab = {'<UNK>':0}
```

```python
def build_vocab(row):
  print(row['question'], row['answer'])
  tokenized_question = tokenize(row['question'])
  tokenized_answer= tokenize(row['answer'])

  merged_tokens = tokenized_question + tokenized_answer

  for token in merged_tokens:
    if token not in vocab:
      vocab[token] = len(vocab)

  print(tokenized_question,tokenized_answer)
```

```python
df.apply(build_vocab,axis = 1)
```

```python
len(vocab)
```

```python
#convert words to numerical indices
```

```python
def text_to_indices(text,vocab):

  indexed_text = []

  for token in tokenize(text):

    if token in vocab:
      indexed_text.append(vocab[token])
    else:
      indexed_text.append(vocab['<UNK>'])

  return indexed_text
```

```python
text_to_indices("Hi how are you doing", vocab)
```

```python
import torch
from torch.utils.data import Dataset,DataLoader
```

```python
class QADataset(Dataset):

  def __init__(self, df, vocab):
    self.df = df
    self.vocab = vocab

  def __len__(self):
    return self.df.shape[0]

  def __getitem__(self, idx):
    raw_question = self.df.iloc[idx]['question']
    raw_answer = self.df.iloc[idx]['answer']

    numerical_question = text_to_indices(raw_question, self.vocab)
    numerical_answer = text_to_indices(raw_answer, self.vocab)

    return torch.tensor(numerical_question), torch.tensor(numerical_answer)
```

```python
dataset = QADataset(df,vocab)
```

```python
dataloader = DataLoader(dataset, batch_size = 1,shuffle = True )
```

```python
for question, answer in dataloader:
  print(question,answer)
```
