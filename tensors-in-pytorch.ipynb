# Mathematical Operations

Scalar Operation


```python
x = torch.rand(2,2)
```

```python
x
```

```python
# addition
x + 2

#subtraction
x-5

#multiplication
x*3

#division
x/3

#int division
(x*100)//3

#mod
((x*100)//3)%2

#power
x**2
```

ElementWise operation


```python
a = torch.rand(2,2)
b = torch.rand(2,2)

print(a)
print(b)
```

```python
#add
torch.add(a,b)

#sub
torch.sub(a,b)

#mul
torch.mul(a,b)

#div
torch.div(a,b)

#power
torch.pow(a,b)

#mod
a%b
```

```python
c = torch.tensor([1,-2,3,-4])
```

```python
#abs
torch.abs(c)
```

```python
#negative
torch.neg(c)
```

```python
d = torch.tensor([1.9,2.3,3.7,4.4])
```

```python
#round
torch.round(d)

#ceil
torch.ceil(d)

#floor
torch.floor(d)

#trunc
torch.trunc(d)

#frac
torch.frac(d)
```

Reduction Operation


```python
e = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float64)
e
```

```python
#sum
torch.sum(e)
#sum along columns
torch.sum(e,dim = 0)
#sum along rows
torch.sum(e,dim = 1)
```

```python
#mean
torch.mean(e)
#mean along columns
torch.mean(e,dim = 0)
#mean along rows
torch.mean(e,dim = 1)
```

```python
#median
torch.median(e)
#median along columns
torch.median(e,dim = 0)
#median along rows
torch.median(e,dim = 1)
```

```python
#max and min

torch.min(e)
torch.min(e,dim = 0)
torch.min(e,dim = 1)

torch.max(e)
torch.max(e,dim = 0)
torch.max(e,dim = 1)
```

```python
#product

torch.prod(e)
```

```python
#std dev
torch.std(e)
```

```python
#variance
torch.var(e)
```

```python
#argmax
torch.argmax(e)
```

Matrix Operation


```python
f = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float64)
g = torch.randint(size=(3,2), low = 0 ,high = 10,dtype=torch.float64)

print(f)
print(g)
```

```python
#matrix multiplication
torch.matmul(f,g)
```

```python
vector1 = torch.tensor([1,2])
vector2 = torch.tensor([3,4])
```

```python
#dot product
torch.dot(vector1,vector2)
```

```python
#transpose
f
```

```python
torch.transpose(f,0,1)
```

```python
h = torch.randint(size= (3,3), low = 0 , high = 10,dtype = torch.float32)
```

```python
#determinant
torch.det(h)
```

```python
#inverse
torch.inverse(h)
```

Comparison operation

```python
i = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float64)
j = torch.randint(size=(2,3), low=0, high=10, dtype=torch.float64)

print(i)
print(j)
```

```python
#greater than
i>j
```

```python
#less than
i<j
```
